```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split

# 1. Load the SMS Spam Collection dataset directly from a public repository
url = "https://raw.githubusercontent.com/mohitgupta-1O1/Kaggle-SMS-Spam-Collection-Dataset-/master/spam.csv"
df = pd.read_csv(url, encoding="latin-1")

# Keep only the label and text columns
df = df[['v1', 'v2']]
df.columns = ['label', 'text']

emails = df['text']
labels = df['label']

# 2. Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    emails, labels, test_size=0.2, random_state=42
)

# 3. The 10-Line Machine Learning Model
vec = TfidfVectorizer()                    # Convert text -> numbers (features)
model = MultinomialNB()                    # A simple, fast classifier
model.fit(vec.fit_transform(X_train), y_train) # Learn from training data

# Check baseline accuracy
accuracy = model.score(vec.transform(X_test), y_test)
print(f"Baseline Model Accuracy: {accuracy * 100:.2f}%")
```

```python
# 5 tricky test messages (mix of spam that might slip through and normal mail that might get flagged)
tests = [
    "Cl@im your fr33 g1ft card now",       # Spam with leetspeak/symbols
    "Not free, just confirming lunch at 1pm", # Normal mail with the word "free"
    "Urgent! Claim your cash prize via link", # Standard spam style
    "Hey bro, are we still free to hang out later?", # Normal mail with "free"
    "W0n 1,000,000 dollars click h3re"       # Obvious spam with numbers/symbols
]

# Run predictions and print what the model thinks
predictions = model.predict(vec.transform(tests))

for message, pred in zip(tests, predictions):
    print(f"Message: '{message}'\n--> Model Predicts: {pred.upper()}\n")
```
